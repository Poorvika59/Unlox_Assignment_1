Q11.

[[99, 0, 0], [0, 0, 0], [0, 0, 0]]

[[99, 0, 0], [99, 0, 0], [99, 0, 0]]

In matrix1 the comprehension biulds three separate inner lists. In matrix2,[[0]*3] *3 copies the reference to the same inner list three times, so all rows one object and changing one row changes all of them.

```python
#Q11.
matrix1 = [[0] * 3 for _ in range(3)]
matrix2 = [[0] * 3] * 3
matrix1[0][0] = 99
matrix2[0][0] = 99
print(matrix1)
print(matrix2)
```

Q12.

x = [1, 2, 3, 4]

y = [1, 2, 3]

a = [1, 2, 3, 4]

b = [1, 2, 3, 4]

x = x + [4] builds a new list and rebinds x to it, so y still points to the old list. a += [4] calls an in-place extend on the same list objject, so b sees the change too


```python
#Q12
x = [1, 2, 3]
y = x
x = x + [4]
print("x =", x)
print("y =", y)

a = [1, 2, 3]
b = a
a += [4]
print("a =", a)
print("b =", b)
```
